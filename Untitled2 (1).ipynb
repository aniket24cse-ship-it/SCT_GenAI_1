```python
import os
import numpy as np
import pandas as pd
from pypdf import PdfReader
from rank_bm25 import BM25Okapi
from sentence_transformers import SentenceTransformer, CrossEncoder
import faiss
from groq import Groq

# 1. Tumhari Groq API Key aur PDF Path
API_KEY = "YOUR_API_KEY"
PDF_PATH = "/content/research.pdf"

# 2. Groq Client & Auto-Select Available Model
client = Groq(api_key=API_KEY)
available_models = [
    m.id for m in client.models.list().data
    if not any(x in m.id.lower() for x in ["whisper", "guard", "tts", "orpheus"])
]
SELECTED_MODEL = available_models[0]
print(f"🤖 Active Groq Model Selected: {SELECTED_MODEL}")

# 3. PDF Read & Chunking (Page number ke saath)
reader = PdfReader(PDF_PATH)
chunks = []
for page_idx, page in enumerate(reader.pages):
    text = page.extract_text() or ""
    for i in range(0, len(text), 500):
        piece = text[i:i+600].strip()
        if piece:
            chunks.append({"page": page_idx + 1, "text": piece})

print(f"✅ PDF Loaded! Total {len(reader.pages)} pages se {len(chunks)} chunks bane.\n")

# 4. Lexical Retrieval: BM25 (Keyword Search)
tokenized_corpus = [c["text"].lower().split() for c in chunks]
bm25 = BM25Okapi(tokenized_corpus)

# 5. Dense Retrieval: FAISS Vector Embeddings
embed_model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = embed_model.encode([c["text"] for c in chunks], convert_to_numpy=True)
faiss.normalize_L2(embeddings)
index = faiss.IndexFlatIP(embeddings.shape[1])
index.add(embeddings)

# 6. Cross-Encoder Reranker (BAAI/bge-reranker-base)
reranker = CrossEncoder("BAAI/bge-reranker-base")

def ask_question(query):
    print("=" * 65)
    print(f"❓ SAWAL: {query}")
    print("=" * 65)

    # BM25 Top 5
    bm25_scores = bm25.get_scores(query.lower().split())
    bm25_top_idx = np.argsort(bm25_scores)[::-1][:5]

    # FAISS Top 5
    q_emb = embed_model.encode([query], convert_to_numpy=True)
    faiss.normalize_L2(q_emb)
    _, faiss_top_idx = index.search(q_emb, 5)
    faiss_top_idx = faiss_top_idx[0]

    # Hybrid Retrieval (BM25 + FAISS combined)
    hybrid_idx = list(dict.fromkeys(list(bm25_top_idx) + list(faiss_top_idx)))
    candidates = [chunks[i] for i in hybrid_idx]

    # Reranker se Top 3 filter karna
    pairs = [[query, c["text"]] for c in candidates]
    rerank_scores = reranker.predict(pairs)
    top3_indices = np.argsort(rerank_scores)[::-1][:3]
    top3_docs = [candidates[i] for i in top3_indices]

    # Answer generation with Source Citations
    context = "\n".join([f"[Page {d['page']}]: {d['text']}" for d in top3_docs])
    prompt = f"""Answer the question strictly using the provided context.
Include exact page number citations like [Page X] after every fact.

Context:
{context}

Question: {query}"""

    chat = client.chat.completions.create(
        messages=[{"role": "user", "content": prompt}],
        model=SELECTED_MODEL,
        temperature=0
    )

    print("\n💡 JAWAB (WITH SOURCE CITATIONS):")
    print(chat.choices[0].message.content)

    # Deliverable: Retrieval Comparison (BM25 vs. Dense vs. Hybrid)
    df = pd.DataFrame({
        "Retrieval Method": ["BM25 (Lexical Keyword)", "Dense (FAISS Vector)", "Hybrid + Reranker (Final)"],
        "Top 3 Retrieved Pages": [
            [chunks[i]["page"] for i in bm25_top_idx[:3]],
            [chunks[i]["page"] for i in faiss_top_idx[:3]],
            [d["page"] for d in top3_docs]
        ]
    })
    print("\n📊 RETRIEVAL COMPARISON (BM25 vs. Dense vs. Hybrid):")
    print(df.to_string(index=False))
    print("\n")

# --- Is Research Paper par 3 Test Questions ---
ask_question("How many students participated in this study and what was their average age?")
ask_question("Which mobile apps and online resources did students use most frequently?")
ask_question("What are the main limitations of this study mentioned by the author?")
```
